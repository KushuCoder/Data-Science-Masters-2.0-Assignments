Q1. Calculate the 95% confidence interval for a sample of data with a mean of 50 and a standard deviation
of 5 using Python. Interpret the results.

```python
import scipy.stats as stats
import math
mean = 50
std_dev = 5
n = 30
confidence_level = 0.95
standard_error = std_dev / math.sqrt(n)
z_critical = stats.norm.ppf(1 - (1 - confidence_level) / 2)
margin_of_error = z_critical * standard_error
confidence_interval = (mean - margin_of_error, mean + margin_of_error)
confidence_interval
```

Interpretation: The 95% confidence interval for the true population mean is calculated based on the sample data. This means that we are 95% confident that the true population mean lies within the calculated range. The result from the Python code will give the lower and upper bounds of this confidence interval.

Q2. Conduct a chi-square goodness of fit test to determine if the distribution of colors of M&Ms in a bag
matches the expected distribution of 20% blue, 20% orange, 20% green, 10% yellow, 10% red, and 20%
brown. Use Python to perform the test with a significance level of 0.05.

```python
import scipy.stats as stats
import numpy as np
observed = np.array([25, 30, 20, 15, 10, 20])
total_MMs = observed.sum()
expected_percentages = np.array([0.20, 0.20, 0.20, 0.10, 0.10, 0.20])
expected = expected_percentages * total_MMs
chi2_stat, p_value = stats.chisquare(observed, expected)
chi2_stat, p_value
```

Interpretation:

The chi-square statistic (chi2_stat) tells us how much the observed frequencies deviate from the expected frequencies.

The p-value (p_value) will help determine if the observed distribution significantly differs from the expected distribution. If the p-value is less than the significance level (0.05), we reject the null hypothesis, indicating that the distribution does not match the expected distribution. If the p-value is greater than 0.05, we fail to reject the null hypothesis, meaning there is no significant difference between the observed and expected distributions.

Q3. Use Python to calculate the chi-square statistic and p-value for a contingency table with the following
data:

|           | Group A | Group B |
|-----------|---------|---------|
| Outcome 1 | 20      | 15      |
| Outcome 2 | 10      | 25      |
| Outcome 3 | 15      | 20      |

Interpret the results of the test.

```python
import scipy.stats as stats
import numpy as np
observed = np.array([[20, 15],
                     [10, 25],
                     [15, 20]])
chi2_stat, p_value, dof, expected = stats.chi2_contingency(observed)
chi2_stat, p_value, dof, expected
```

Interpretation of Results:
Chi-square statistic: This value tells us how much the observed frequencies differ from the expected frequencies. A higher value indicates a larger discrepancy.

P-value: This value helps us determine whether the differences observed are statistically significant. A smaller p-value (typically < 0.05) suggests that the differences are statistically significant, meaning the groups are likely to differ in some way.

Degrees of Freedom: This is the number of independent comparisons you can make, calculated as (number of rows - 1) * (number of columns - 1).

Expected Frequencies: These values are the frequencies we'd expect to observe if there were no relationship between the groups and outcomes. Comparing the observed values to these expected values helps in determining the chi-square statistic.

Q4. A study of the prevalence of smoking in a population of 500 individuals found that 60 individuals
smoked. Use Python to calculate the 95% confidence interval for the true proportion of individuals in the
population who smoke.

```python
import scipy.stats as stats
import numpy as np
n = 500  
x = 60   
p_hat = x / n  
z = stats.norm.ppf(0.975)  
se = np.sqrt((p_hat * (1 - p_hat)) / n)
lower_bound = p_hat - z * se
upper_bound = p_hat + z * se
(lower_bound, upper_bound)
```

Q5. Calculate the 90% confidence interval for a sample of data with a mean of 75 and a standard deviation
of 12 using Python. Interpret the results.

```python
import scipy.stats as stats
import numpy as np
mean = 75  
std_dev = 12  
n = 30  
z = stats.norm.ppf(0.95)  
se = std_dev / np.sqrt(n)
lower_bound = mean - z * se
upper_bound = mean + z * se
(lower_bound, upper_bound)
```

Interpretation:
The 90% confidence interval means that we are 90% confident that the true population mean lies within the calculated range. If we repeatedly sample and compute confidence intervals in the same way, 90% of those intervals would contain the true mean.


Q6. Use Python to plot the chi-square distribution with 10 degrees of freedom. Label the axes and shade the
area corresponding to a chi-square statistic of 15.

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
df = 10  
x = np.linspace(0, 30, 500)  
y = stats.chi2.pdf(x, df)
plt.figure(figsize=(8, 5))
plt.plot(x, y, label=r'$\chi^2$ Distribution (df=10)', color='blue')
x_shade = np.linspace(15, 30, 200)
y_shade = stats.chi2.pdf(x_shade, df)
plt.fill_between(x_shade, y_shade, color='red', alpha=0.5, label='Shaded Area (χ² > 15)')
plt.axvline(15, color='red', linestyle='--', label='χ² = 15')
plt.xlabel('Chi-Square Statistic')
plt.ylabel('Probability Density')
plt.title(r'Chi-Square Distribution with 10 Degrees of Freedom')
plt.legend()
plt.grid(True)
plt.show()
```

Q7. A random sample of 1000 people was asked if they preferred Coke or Pepsi. Of the sample, 520
preferred Coke. Calculate a 99% confidence interval for the true proportion of people in the population who
prefer Coke.

```python
import scipy.stats as stats
import numpy as np
n = 1000  
x = 520   
p_hat = x / n  
z = stats.norm.ppf(0.995)  
se = np.sqrt((p_hat * (1 - p_hat)) / n)
lower_bound = p_hat - z * se
upper_bound = p_hat + z * se
(lower_bound, upper_bound)
```

Q8. A researcher hypothesizes that a coin is biased towards tails. They flip the coin 100 times and observe
45 tails. Conduct a chi-square goodness of fit test to determine if the observed frequencies match the
expected frequencies of a fair coin. Use a significance level of 0.05.

```python
import scipy.stats as stats
import numpy as np
observed = np.array([45, 55])
expected = np.array([50, 50])
chi2_stat, p_value = stats.chisquare(observed, expected)
chi2_stat, p_value
```

Interpretation:
-If p-value < 0.05, reject the null hypothesis: The coin is biased.
-If p-value ≥ 0.05, fail to reject the null hypothesis: No sufficient evidence to say the coin is biased.


Q9. A study was conducted to determine if there is an association between smoking status (smoker or
non-smoker) and lung cancer diagnosis (yes or no). The results are shown in the contingency table below.
Conduct a chi-square test for independence to determine if there is a significant association between
smoking status and lung cancer diagnosis.

|             | Lung Cancer: Yes |Lung Cancer: No  |
|-------------|------------------|-----------------|
| Smoker      |       60         |       140       |
| Non-smoker  |       30         |       170       |

Use a significance level of 0.05.


Chi-Square Test for Independence

Hypotheses:
-Null Hypothesis (H₀): Smoking status and lung cancer diagnosis are independent (no association).
-Alternative Hypothesis (H₁): Smoking status and lung cancer diagnosis are dependent (there is an association).

```python
import scipy.stats as stats
import numpy as np
observed = np.array([[60, 140], 
                     [30, 170]])

chi2_stat, p_value, dof, expected = stats.chi2_contingency(observed)
chi2_stat, p_value
```

Interpretation:
-If p-value < 0.05, reject the null hypothesis: There is a significant association between smoking status and lung cancer.
-If p-value ≥ 0.05, fail to reject the null hypothesis: No sufficient evidence to suggest an association.

Q11. A random sample of 30 people was selected from a population with an unknown mean and standard
deviation. The sample mean was found to be 72 and the sample standard deviation was found to be 10.
Conduct a hypothesis test to determine if the population mean is significantly different from 70. Use a
significance level of 0.05.

One-Sample t-Test for Population Mean

Hypotheses:
-Null Hypothesis (H₀): The population mean is equal to 70 
-Alternative Hypothesis (H₁): The population mean is different from 70 .

Test Statistic:
Since the population standard deviation is unknown, we use the t-test


```python
import scipy.stats as stats
import numpy as np
sample_mean = 72
hypothesized_mean = 70
sample_std = 10
n = 30
t_stat = (sample_mean - hypothesized_mean) / (sample_std / np.sqrt(n))
p_value = 2 * (1 - stats.t.cdf(abs(t_stat), df=n-1))
t_stat, p_value
```

Interpretation:
-If p-value < 0.05, reject the null hypothesis: 
The population mean is significantly different from 70.
-If p-value ≥ 0.05, fail to reject the null hypothesis: No sufficient evidence to conclude a significant difference.
