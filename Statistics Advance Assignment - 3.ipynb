Q1: What is Estimation Statistics? Explain point estimate and interval estimate.


Estimation Statistics is the process of inferring or estimating the population parameters (such as mean, variance) based on sample data. The two main types of estimation are:

1.Point Estimate: A point estimate is a single value that is used to estimate a population parameter. It is typically the sample mean or sample proportion.
   
2.Interval Estimate: An interval estimate provides a range of values within which the population parameter is likely to lie. This is typically expressed as a confidence interval, such as a 95% confidence interval.

Q2. Write a Python function to estimate the population mean using a sample mean and standard deviation.


```python
import math
def estimate_population_mean(sample_mean, sample_std, sample_size, confidence_level):
    z_scores = {0.90: 1.645, 0.95: 1.96, 0.99: 2.576}
    z = z_scores.get(confidence_level, 1.96)
    margin_of_error = z * (sample_std / math.sqrt(sample_size))
    lower_bound = sample_mean - margin_of_error
    upper_bound = sample_mean + margin_of_error
    return lower_bound, upper_bound
sample_mean = 500
sample_std = 50
sample_size = 50
confidence_level = 0.95
lower, upper = estimate_population_mean(sample_mean, sample_std, sample_size, confidence_level)
print(f"The estimated population mean is between {lower} and {upper}.")
```

Q3: What is Hypothesis testing? Why is it used? State the importance of Hypothesis testing.


Hypothesis testing is a statistical method used to make inferences or draw conclusions about a population based on sample data. It involves two competing hypotheses:

1.Null Hypothesis (H0): The statement that there is no effect or no difference.
2.Alternative Hypothesis (H1): The statement that there is a significant effect or difference.

Purpose: It is used to determine whether there is enough evidence to reject the null hypothesis. Hypothesis testing is used in decision-making to validate or refute claims based on sample data.

Importance:
1.It allows us to make evidence-based decisions.
2.It quantifies uncertainty and the likelihood of an outcome occurring.
3.It helps validate scientific and research-based claims.

Q5. Write a Python script to conduct a hypothesis test on the difference between two population means, given a sample from each population.


```python
import scipy.stats as stats
def two_sample_t_test(sample1, sample2):
    mean1, mean2 = sample1.mean(), sample2.mean()
    std1, std2 = sample1.std(), sample2.std()
    n1, n2 = len(sample1), len(sample2)
    pooled_std = math.sqrt(((std1**2) / n1) + ((std2**2) / n2))
    t_stat = (mean1 - mean2) / pooled_std
    df = min(n1 - 1, n2 - 1)
    p_value = 2 * (1 - stats.t.cdf(abs(t_stat), df))
    return t_stat, p_value
import numpy as np
sample1 = np.array([100, 102, 105, 107, 110])
sample2 = np.array([98, 101, 104, 106, 108])
t_stat, p_value = two_sample_t_test(sample1, sample2)
print(f"T-statistic: {t_stat}, P-value: {p_value}")
```

Q6: What is a null and alternative hypothesis? Give some examples.


The Null Hypothesis (H0) is a statement that suggests no effect, difference, or relationship exists. It represents the idea that any observed difference is due to random chance.

The Alternative Hypothesis (H1) is a statement that suggests the presence of an effect, difference, or relationship. It represents the opposite of the null hypothesis.

Examples:
1.Null Hypothesis: There is no difference in the average salaries of men and women in the company.
   Alternative Hypothesis: There is a difference in the average salaries of men and women in the company.

2.Null Hypothesis: The new medication has no effect on blood pressure.
   Alternative Hypothesis: The new medication lowers blood pressure.


Q7: Write down the steps involved in hypothesis testing.


1.State the hypotheses: Define the null and alternative hypotheses.
2.Choose the significance level (α): Typically set at 0.05, representing a 5% risk of rejecting the null hypothesis when it is true.
3.Select the appropriate test: Choose a test based on the data and research question (e.g., t-test, chi-square test).
4.Compute the test statistic: Use the chosen test to calculate the test statistic.
5.Find the p-value: The p-value is the probability of obtaining the observed results under the null hypothesis.
6.Make a decision: Compare the p-value with the significance level α:
   -If p-value ≤ α, reject the null hypothesis.
   -If p-value > α, fail to reject the null hypothesis.
7.Conclusion: State the result of the hypothesis test.


Q8: Define p-value and explain its significance in hypothesis testing.


The p-value is the probability of obtaining test results at least as extreme as the results actually observed, under the assumption that the null hypothesis is true.
Significance in Hypothesis Testing:
-If the p-value is less than or equal to the chosen significance level (α), it suggests that the observed data is inconsistent with the null hypothesis, leading to rejection of the null hypothesis.
-A smaller p-value indicates stronger evidence against the null hypothesis.

For example, if the p-value is 0.03 and α is 0.05, we would reject the null hypothesis, as 0.03 < 0.05.

When to use the t-Distribution:
-When the sample size is small and the population standard deviation is unknown.
-When performing hypothesis tests on means with small sample sizes.


Q12: What is t-statistic? State the formula for t-statistic.


The t-statistic is a ratio that compares the difference between the sample mean and the population mean relative to the variation in the sample. It is used in hypothesis testing to determine whether to reject the null hypothesis.

Formula:
t = (X̄ - μ) / (s / √n)

Where:
-X̄ is the sample mean,
-μ is the population mean,
-s is the sample standard deviation,
-n is the sample size.


Q13: A coffee shop owner wants to estimate the average daily revenue for their shop. They take a random sample of 50 days and find the sample mean revenue to be $500 with a standard deviation of $50. Estimate the population mean revenue with a 95% confidence interval.


```python
import math
sample_mean = 500
sample_std = 50
sample_size = 50
confidence_level = 0.95
z_scores = {0.90: 1.645, 0.95: 1.96, 0.99: 2.576}
z = z_scores[confidence_level]
margin_of_error = z * (sample_std / math.sqrt(sample_size))
lower_bound = sample_mean - margin_of_error
upper_bound = sample_mean + margin_of_error
print(f"The estimated population mean revenue is between ${lower_bound} and ${upper_bound}.")
```

Q14: A researcher hypothesizes that a new drug will decrease blood pressure by 10 mmHg. They conduct a clinical trial with 100 patients and find that the sample mean decrease in blood pressure is 8 mmHg with a standard deviation of 3 mmHg. Test the hypothesis with a significance level of 0.05.


```python
import scipy.stats as stats
import math
sample_mean = 8
population_mean = 10
sample_std = 3
sample_size = 100
alpha = 0.05
t_stat = (sample_mean - population_mean) / (sample_std / math.sqrt(sample_size))
df = sample_size - 1
p_value = 2 * (1 - stats.t.cdf(abs(t_stat), df))
if p_value < alpha:
    print(f"Reject the null hypothesis. P-value: {p_value}")
else:
    print(f"Fail to reject the null hypothesis. P-value: {p_value}")
```
