Q1: What are the Probability Mass Function (PMF) and Probability Density Function (PDF)? Explain with an example.


The Probability Mass Function (PMF) is used to describe the probability distribution of a discrete random variable. It gives the probability that a discrete random variable is exactly equal to a certain value. The sum of all probabilities in the PMF must be 1.

The Probability Density Function (PDF) is used for continuous random variables. It describes the likelihood of a random variable taking on a specific value within a given range. The area under the curve of the PDF for a certain range represents the probability that the random variable falls within that range. Unlike PMF, PDF doesn't give the exact probability of a specific value but rather the probability density.

Example:
For a discrete random variable like the roll of a fair six-sided die, the PMF would give the probability of rolling a 1, 2, 3, 4, 5, or 6, each with a probability of 1/6.

For a continuous random variable like the height of adults in a population, the PDF would describe the likelihood of someone having a height between, say, 170 cm and 180 cm.


Q2: What is Cumulative Density Function (CDF)? Explain with an example. Why CDF is used?


The Cumulative Density Function (CDF) of a random variable is the probability that the random variable takes a value less than or equal to a specific value. It is a non-decreasing function that starts from 0 and approaches 1 as the random variable increases.

The CDF can be used for both discrete and continuous random variables. It gives us the cumulative probability up to a certain value, providing a cumulative perspective of the probability distribution.

Example:
For a discrete random variable representing the roll of a fair six-sided die, the CDF at value 4 would be the sum of the probabilities of rolling 1, 2, 3, or 4. In this case, the CDF at 4 would be 4/6 = 2/3, as there is a 2/3 chance of rolling a 4 or less.

CDF is used because it gives a convenient way to compute probabilities for intervals. It also allows us to visualize the distribution and compare probabilities across different ranges.


Q3: What are some examples of situations where the normal distribution might be used as a model? Explain how the parameters of the normal distribution relate to the shape of the distribution.


The normal distribution is commonly used to model situations where the data is symmetrically distributed around the mean. It is applicable in many fields, such as finance, medicine, engineering, and natural sciences.

Examples:
1.Heights of individuals in a population.
2.Test scores in standardized exams.
3.Measurement errors in experiments.
4.Stock market returns.

The normal distribution is characterized by two parameters:
1.Mean (μ): The mean determines the center of the distribution. It is the peak of the bell curve.
2.Standard deviation (σ): The standard deviation controls the spread of the distribution. A larger standard deviation means the data is more spread out, while a smaller standard deviation means the data is more concentrated around the mean.

The normal distribution is symmetric, and the area under the curve equals 1, representing the total probability.

Q5: What is Bernoulli Distribution? Give an Example. What is the difference between Bernoulli Distribution and Binomial Distribution?


The Bernoulli distribution is a discrete probability distribution that models a random experiment with two possible outcomes: success (usually coded as 1) or failure (usually coded as 0). The Bernoulli distribution is parameterized by a single parameter, p, which is the probability of success.

Example:
A coin toss is a Bernoulli experiment. If the coin lands heads, the outcome is a success (1); if it lands tails, the outcome is a failure (0). If the coin is fair, then p = 0.5.

Difference between Bernoulli Distribution and Binomial Distribution:
The Bernoulli distribution models a single trial with two outcomes.
The Binomial distribution models the number of successes in a fixed number of independent Bernoulli trials.
The Binomial distribution has two parameters: n (the number of trials) and p (the probability of success in each trial).


Q6: Consider a dataset with a mean of 50 and a standard deviation of 10. If we assume that the dataset is normally distributed, what is the probability that a randomly selected observation will be greater than 60? Use the appropriate formula and show your calculations.


```python
import scipy.stats as stats
mean = 50
std_dev = 10
value = 60
z_score = (value - mean) / std_dev
probability = 1 - stats.norm.cdf(z_score)
probability
```

Q7: Explain uniform Distribution with an example.


The uniform distribution is a type of probability distribution where all outcomes are equally likely within a certain range. The distribution is defined by two parameters: the minimum (a) and the maximum (b) values, and it is characterized by a flat probability density function.

Example:
If you randomly pick a number between 1 and 10 (inclusive), each number (1, 2, 3, ..., 10) has an equal probability of being selected. This is a discrete uniform distribution.

For a continuous uniform distribution, an example could be the time of day a bus arrives at a station, where the bus is equally likely to arrive at any time between 8:00 AM and 9:00 AM.


Q8: What is the z score? State the importance of the z score.


The z-score is a statistical measurement that describes how many standard deviations a data point is from the mean. It is calculated using the formula:

z = (X - μ) / σ

Where:
X is the data point.
μ is the mean of the distribution.
σ is the standard deviation of the distribution.

The z-score is important because it allows us to compare different data points from different distributions. It helps to understand how unusual or typical a data point is in the context of the distribution. A z-score of 0 means the data point is exactly at the mean, while a z-score of 2 indicates the data point is two standard deviations above the mean.


Q9: What is Central Limit Theorem? State the significance of the Central Limit Theorem.


The Central Limit Theorem (CLT) states that the distribution of the sample mean (or sum) of a large number of independent, identically distributed random variables will approach a normal distribution, regardless of the original distribution of the variables. This approximation becomes more accurate as the sample size increases.

Significance:
The CLT allows us to make inferences about the population mean from sample data, even if the population distribution is not normal.
It justifies the use of normal distribution in hypothesis testing and confidence intervals, even when the underlying data does not follow a normal distribution.


Q10: State the assumptions of the Central Limit Theorem.


The assumptions of the Central Limit Theorem are:
1.The sample observations must be independent of each other.
2.The samples should be drawn from the same population with the same distribution (i.i.d).
3.The sample size should be sufficiently large (usually n > 30 is considered large enough).
4.The underlying distribution of the population can be any distribution, but the sample size must be large for the CLT to hold.
