Q1. Write a Python function that takes in two arrays of data and calculates the F-value for a variance ratio test. The function should return the F-value and the corresponding p-value for the test.


The F-value for a variance ratio test is calculated as the ratio of the variances of two datasets. The corresponding p-value is obtained using the F-distribution. The function takes two arrays as input, calculates their variances, and returns the F-value and p-value.


```python
import numpy as np
from scipy.stats import f
def variance_ratio_test(data1, data2):
    var1 = np.var(data1, ddof=1)
    var2 = np.var(data2, ddof=1)
    f_value = var1 / var2 if var1 > var2 else var2 / var1
    df1 = len(data1) - 1
    df2 = len(data2) - 1
    p_value = 1 - f.cdf(f_value, df1, df2)
    return f_value, p_value
data1 = [10, 12, 14, 16, 18]
data2 = [8, 9, 11, 13, 15]
variance_ratio_test(data1, data2)
```

Q2. Given a significance level of 0.05 and the degrees of freedom for the numerator and denominator of an F-distribution, write a Python function that returns the critical F-value for a two-tailed test.


 The critical F-value for a two-tailed test at a given significance level is determined using the inverse cumulative distribution function of the F-distribution. The function takes the significance level and degrees of freedom as input and returns the critical value.


```python
from scipy.stats import f
def critical_f_value(alpha, df1, df2):
    return f.ppf(1 - alpha / 2, df1, df2)
critical_f_value(0.05, 10, 20)
```

Q3. Write a Python program that generates random samples from two normal distributions with known variances and uses an F-test to determine if the variances are equal. The program should output the F-value, degrees of freedom, and p-value for the test.


This program generates random samples from two normal distributions with known variances and performs an F-test to determine if the variances are equal. The output includes the F-value, degrees of freedom, and p-value.


```python
import numpy as np
from scipy.stats import f
np.random.seed(42)
def f_test_random_samples(size1, size2, var1, var2):
    sample1 = np.random.normal(0, np.sqrt(var1), size1)
    sample2 = np.random.normal(0, np.sqrt(var2), size2)
    var_sample1 = np.var(sample1, ddof=1)
    var_sample2 = np.var(sample2, ddof=1)
    f_value = var_sample1 / var_sample2 if var_sample1 > var_sample2 else var_sample2 / var_sample1
    df1 = size1 - 1
    df2 = size2 - 1
    p_value = 1 - f.cdf(f_value, df1, df2)
    return f_value, df1, df2, p_value
f_test_random_samples(30, 30, 10, 15)
```

Q4. The variances of two populations are known to be 10 and 15. A sample of 12 observations is taken from each population. Conduct an F-test at the 5% significance level to determine if the variances are significantly different.


 To test if the variances of two populations are significantly different, an F-test is conducted using the sample sizes and known variances. The F-value is compared against the critical value at a 5% significance level.


```python
from scipy.stats import f
var1, var2 = 10, 15
n1, n2 = 12, 12
f_value = var1 / var2
df1, df2 = n1 - 1, n2 - 1
p_value = 1 - f.cdf(f_value, df1, df2)
f_value, p_value
```

Q6. Write a Python function that takes in the degrees of freedom for the numerator and denominator of an F-distribution and calculates the mean and variance of the distribution. The function should return the mean and variance as a tuple.


The mean and variance of an F-distribution are calculated using its degrees of freedom. The mean is given by df2 / (df2 - 2) for df2 > 2, and the variance by a more complex formula.


```python
def f_distribution_stats(df1, df2):
    if df2 > 2:
        mean = df2 / (df2 - 2)
    else:
        mean = None
    if df2 > 4:
        variance = (2 * (df2**2) * (df1 + df2 - 2)) / (df1 * ((df2 - 2)**2) * (df2 - 4))
    else:
        variance = None
    return mean, variance
f_distribution_stats(10, 20)
```
