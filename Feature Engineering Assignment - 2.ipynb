Q1. What is the Filter method in feature selection, and how does it work?


The Filter method is a feature selection technique that evaluates the relevance of input features based on statistical measures rather than training a machine learning model. It works by independently ranking features based on their correlation with the target variable.     

How It Works:     
Compute a statistical measure (e.g., correlation, chi-square test, mutual information).    
Rank features based on their scores.     
Select top-ranked features and discard the irrelevant ones.

Q3. What are some common techniques used in Embedded feature selection methods?


Embedded methods integrate feature selection within the training process of machine learning models. These techniques automatically select the most important features during model training.     

Common Techniques:     
Lasso (L1 Regularization) – Shrinks some feature coefficients to zero, effectively removing them.     
Decision Tree-Based Feature Importance – Uses tree-based models (Random Forest, XGBoost) to rank features by importance.     
Recursive Feature Elimination (RFE) with Regularization – Eliminates least important features iteratively.

Q4. What are some drawbacks of using the Filter method for feature selection?


Ignores Feature Interactions: Evaluates features independently, not considering how they interact.     
May Remove Important Features: Some low-correlation features might still be useful in combination.     
Not Optimized for a Specific Model: Selected features may not be the best choice for a particular ML model.     
Risk of Overlooking Nonlinear Relationships: Only captures linear dependencies (e.g., Pearson correlation).

Q5. In which situations would you prefer using the Filter method over the Wrapper method for feature selection?


When working with large datasets – The Filter method is computationally efficient.     
As a preprocessing step – Reduces dimensionality before applying more complex methods.     
When using algorithms robust to irrelevant features – Some models like Random Forest handle redundant features well.     
When interpretability is needed – Provides clear insights into feature importance.

Q7. You are working on a project to predict the outcome of a soccer match. You have a large dataset with
many features, including player statistics and team rankings. Explain how you would use the Embedded
method to select the most relevant features for the model.


Train a Decision Tree-based model (e.g., XGBoost, Random Forest).     
Extract feature importance scores from the trained model.     
Keep only the top-ranked features based on importance.


```python
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
import pandas as pd
data = load_iris()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = data.target
rf = RandomForestClassifier()
rf.fit(X, y)
importances = rf.feature_importances_
selected_features = X.columns[importances > 0.02]
print("Selected Features:", selected_features.tolist())
```
