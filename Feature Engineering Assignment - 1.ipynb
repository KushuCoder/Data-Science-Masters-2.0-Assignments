Q1: What are missing values in a dataset? Why is it essential to handle missing values? Name some
algorithms that are not affected by missing values.


Missing values occur in a dataset when no data is stored for a specific feature or observation. They can arise due to human errors, data corruption, or technical issues during data collection.     

Importance of Handling Missing Values:    

Prevents biased results in model training.    
Enhances the accuracy and reliability of predictions.     
Ensures data consistency and completeness for meaningful analysis.    
Algorithms not affected by missing values:    

Decision Trees (e.g., CART, Random Forest) handle missing values internally.     
k-Nearest Neighbors (KNN) can handle missing data by imputing based on nearest neighbors.    
XGBoost can handle missing values by learning optimal splits.

Q2: List down techniques used to handle missing data. Give an example of each with python code.


Techniques to Handle Missing Data:  
Removing Missing Data (If missing values are too many)    
Imputation     
Mean/Median/Mode Imputation    
KNN Imputation    
Forward/Backward Fill     
Using Algorithms that Handle Missing Values    
Predicting Missing Values using ML models

```python
import pandas as pd
import numpy as np
data = {'Age': [25, np.nan, 30, 35, np.nan, 40],
        'Salary': [50000, 60000, np.nan, 80000, 90000, 100000]}
df = pd.DataFrame(data)
df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Salary'].fillna(df['Salary'].median(), inplace=True)
print(df)
```

Q3: Explain the imbalanced data. What will happen if imbalanced data is not handled?


Imbalanced data occurs when one class in a classification problem has significantly more observations than another.    

Effects of not handling imbalanced data:     

The model may be biased toward the majority class.     
Poor generalization to minority class examples.     
Low recall (high false negatives) for the minority class.    
Example: In fraud detection, if 98% of transactions are legitimate and 2% are fraudulent, a model predicting "legitimate" always will still have 98% accuracy but be useless for fraud detection.


Q4: What are Up-sampling and Down-sampling? Explain with an example when up-sampling and down-sampling are required.


Up-sampling (Oversampling) increases the number of minority class samples by duplicating existing instances or generating synthetic ones.     
Down-sampling (Undersampling) reduces the number of majority class samples to balance the dataset.


```python
from imblearn.over_sampling import SMOTE
from collections import Counter
from sklearn.model_selection import train_test_split
X = [[1], [2], [3], [4], [5], [6]]
y = [0, 0, 0, 0, 1, 1]
smote = SMOTE()
X_res, y_res = smote.fit_resample(X, y)

print("Original Class Distribution:", Counter(y))
print("Resampled Class Distribution:", Counter(y_res))
```


Q5: What is data Augmentation? Explain SMOTE.


Data augmentation artificially increases the dataset size by creating modified versions of existing data (e.g., rotating, flipping images in deep learning).    

SMOTE (Synthetic Minority Over-sampling Technique) generates synthetic examples rather than duplicating minority class samples. It creates new samples along the line segments joining existing minority class instances.

Q6: What are outliers in a dataset? Why is it essential to handle outliers?


Outliers are extreme values in a dataset that significantly differ from other observations.    

Importance of Handling Outliers:     

They can skew statistical measures like mean and standard deviation.     
They may lead to poor model performance.     
They can distort relationships between variables.      
Common methods to handle outliers:     

Z-score method (removing points beyond 3 standard deviations).     
IQR method (removing points outside Q1 - 1.5IQR and Q3 + 1.5IQR).     
Transformation techniques (log/square root transformation).
