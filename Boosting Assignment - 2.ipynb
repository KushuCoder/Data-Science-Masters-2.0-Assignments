Q1. What is Gradient Boosting Regression?


Gradient Boosting Regression is an ensemble technique that builds a strong regression model by sequentially adding weak learners, typically decision trees. Each new learner corrects the residual errors of the combined ensemble by optimizing a loss function using gradient descent.


Q2. Implement a simple gradient boosting algorithm from scratch using Python and NumPy. Use a simple regression problem as an example and train the model on a small dataset. Evaluate the model's performance using metrics such as mean squared error and R-squared.


```python
import numpy as np
from sklearn.metrics import mean_squared_error, r2_score
X = np.array([1, 2, 3, 4, 5])
y = np.array([1.2, 1.9, 3.2, 3.9, 5.1])
n_estimators = 10
learning_rate = 0.1
pred = np.full_like(y, fill_value=np.mean(y), dtype=np.float64)
models = []
for i in range(n_estimators):
    residuals = y - pred
    gradient = residuals

    best_split = np.mean(X)
    left_mask = X <= best_split
    right_mask = X > best_split

    left_value = np.mean(gradient[left_mask])
    right_value = np.mean(gradient[right_mask])

    stump_pred = np.where(left_mask, left_value, right_value)
    pred += learning_rate * stump_pred
    models.append((best_split, left_value, right_value))
mse = mean_squared_error(y, pred)
r2 = r2_score(y, pred)
print("Mean Squared Error:", mse)
print("R-squared:", r2)
```

Q3. Experiment with different hyperparameters such as learning rate, number of trees, and tree depth to optimise the performance of the model. Use grid search or random search to find the best hyperparameters.


```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import GridSearchCV
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([1.2, 1.9, 3.2, 3.9, 5.1])
param_grid = {
    'n_estimators': [10, 50, 100],
    'learning_rate': [0.01, 0.1, 0.5],
    'max_depth': [1, 2, 3]
}
model = GradientBoostingRegressor()
grid_search = GridSearchCV(model, param_grid, cv=3, scoring='neg_mean_squared_error')
grid_search.fit(X, y)
print("Best Parameters:", grid_search.best_params_)
```

Q4. What is a weak learner in Gradient Boosting?


A weak learner in Gradient Boosting is typically a decision tree with limited depth (often referred to as a decision stump). It is a model that performs only slightly better than random guessing, and it is improved upon iteratively in the boosting process.


Q5. What is the intuition behind the Gradient Boosting algorithm?


The intuition behind Gradient Boosting is to minimize a loss function by iteratively adding models that correct the errors made by the existing ensemble. Each model is trained on the residuals (errors) of the previous model, following the negative gradient of the loss function.


Q6. How does Gradient Boosting algorithm build an ensemble of weak learners?


Gradient Boosting builds the ensemble by initializing with a constant prediction and then adding one weak learner at a time. Each new learner is trained to predict the negative gradient (residuals) of the loss function with respect to the current ensemble's predictions.


Q7. What are the steps involved in constructing the mathematical intuition of Gradient Boosting algorithm?


1.Initialize the model with a constant value that minimizes the loss function.
2.For each iteration:
   -Compute the negative gradient (residuals) of the loss function.
   -Fit a weak learner to the negative gradient.
   -Determine the optimal step size (learning rate).
   -Update the model by adding the weighted predictions of the weak learner.
3.Repeat until the maximum number of iterations is reached or improvement becomes negligible.
