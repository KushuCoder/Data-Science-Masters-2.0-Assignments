Q1. What is an ensemble technique in machine learning?


Ensemble techniques in machine learning combine predictions from multiple models to improve overall performance. The idea is that a group of weak learners can come together to form a strong learner, reducing variance, bias, or improving predictions.

Examples include bagging, boosting, and stacking.


Q2. Why are ensemble techniques used in machine learning?


Ensemble techniques are used to improve the accuracy and robustness of models. They help to:

-Reduce overfitting
-Increase stability
-Improve predictions, especially when individual models are weak

They are especially useful in competitions and real-world applications.

Q3. What is bagging?


Bagging (Bootstrap Aggregating) is an ensemble method that trains multiple instances of the same model on different subsets of the training data, obtained through bootstrapping (random sampling with replacement).

The final prediction is made by averaging (for regression) or voting (for classification).


Q4. What is boosting?


Boosting is an ensemble technique that combines models sequentially, where each new model attempts to correct the errors of the previous one.

Common boosting algorithms include AdaBoost, Gradient Boosting, and XGBoost.


Q5. What are the benefits of using ensemble techniques?


Benefits of using ensemble techniques include:

-Improved accuracy and robustness
-Better generalization to unseen data
-Reduction of model variance and bias
-Ability to handle complex datasets more effectively


Q6. Are ensemble techniques always better than individual models?


No, ensemble techniques are not always better. They can be computationally expensive, harder to interpret, and sometimes overkill for simple problems where a single model performs well.

They also may not improve performance if individual models are already highly accurate and uncorrelated.


Q7. How is the confidence interval calculated using bootstrap?


The confidence interval using bootstrap is calculated by:

1.Generating many resampled datasets from the original data.
2.Calculating the statistic (e.g., mean) for each resample.
3.Taking the desired percentiles (e.g., 2.5th and 97.5th) of the distribution of the statistics to get the 95% confidence interval.

Q8. How does bootstrap work and What are the steps involved in bootstrap?


Bootstrap is a resampling technique used to estimate the distribution of a statistic by sampling with replacement.

Steps involved:

1.Take a sample dataset.
2.Create many bootstrap samples (resample with replacement).
3.Calculate the statistic of interest on each sample.
4.Use the distribution of these statistics to estimate confidence intervals or standard errors.

Q9. A researcher wants to estimate the mean height of a population of trees. They measure the height of a sample of 50 trees and obtain a mean height of 15 meters and a standard deviation of 2 meters. Use bootstrap to estimate the 95% confidence interval for the population mean height.


```python
import numpy as np
np.random.seed(0)
sample_heights = np.random.normal(loc=15, scale=2, size=50)
bootstrap_means = []
for _ in range(1000):
    sample = np.random.choice(sample_heights, size=50, replace=True)
    bootstrap_means.append(np.mean(sample))
lower = np.percentile(bootstrap_means, 2.5)
upper = np.percentile(bootstrap_means, 97.5)
lower, upper
```
