Q1: What is the difference between a t-test and a z-test? Provide an example scenario where you would use each type of test.


A t-test is used when the sample size is small (typically less than 30) or the population standard deviation is unknown. It compares the sample mean to the population mean or compares the means of two groups. A z-test is used when the sample size is large (typically over 30) and the population standard deviation is known.

Example of t-test: A researcher wants to test if the average weight of apples from a new farm is different from 150 grams. Since the sample size is small and the population standard deviation is unknown, a t-test is used.

Example of z-test: A company wants to test if the average height of workers is different from the national average of 5'10". The sample size is large, and the population standard deviation is known, so a z-test is appropriate.


Q2: Differentiate between one-tailed and two-tailed tests.


A one-tailed test tests for the possibility of the relationship in one direction, while a two-tailed test tests for the possibility in both directions.

Example of one-tailed test: A company tests if a new product increases sales above a certain threshold. The hypothesis is that the sales will be higher, not lower.

Example of two-tailed test: A researcher tests if a new drug has an effect on blood pressure, either increasing or decreasing it. The hypothesis tests for both directions of effect.


Q3: Explain the concept of Type 1 and Type 2 errors in hypothesis testing. Provide an example scenario for each type of error.


Type 1 error occurs when a true null hypothesis is incorrectly rejected (false positive), and Type 2 error occurs when a false null hypothesis is incorrectly accepted (false negative).

Example of Type 1 error: A medical test incorrectly indicates that a patient has a disease when they do not (rejecting the null hypothesis that the patient is healthy).
  
Example of Type 2 error: A medical test fails to detect a disease when the patient actually has it (failing to reject the null hypothesis that the patient is healthy).


Q4: Explain Bayes's theorem with an example.


Bayes's theorem calculates the probability of an event based on prior knowledge of conditions that might be related to the event. It is expressed as:

P(A|B) = (P(B|A) * P(A)) / P(B)

Example: If 1% of people have a certain disease, and a test for the disease has a 90% chance of correctly identifying those with the disease and a 5% chance of incorrectly identifying those without it, Bayes's theorem helps calculate the probability that a person has the disease given a positive test result.


Q5: What is a confidence interval? How to calculate the confidence interval, explain with an example.


A confidence interval is a range of values that is used to estimate the true population parameter. It is calculated using the sample mean, the standard error, and a critical value from a statistical distribution.

Formula for confidence interval: 

CI = sample mean ± (critical value * standard error)

Example: Suppose a sample of 100 people has a mean age of 35 years with a standard deviation of 5 years. A 95% confidence interval for the mean age would be calculated as:

CI = 35 ± (1.96 * (5/sqrt(100))) = 35 ± 0.98 = [34.02, 35.98]

Q7: Calculate the 95% confidence interval for a sample of data with a mean of 50 and a standard deviation of 5. Interpret the results.


```python
import scipy.stats as stats
mean = 50
std_dev = 5
n = 30 
confidence_level = 0.95
standard_error = std_dev / (n ** 0.5)
z_critical = stats.norm.ppf(1 - (1 - confidence_level) / 2)
margin_of_error = z_critical * standard_error
confidence_interval = (mean - margin_of_error, mean + margin_of_error)
confidence_interval
```

Q8: What is the margin of error in a confidence interval? How does sample size affect the margin of error? Provide an example of a scenario where a larger sample size would result in a smaller margin of error.


Margin of error is the range within which the true population parameter is expected to lie, based on the sample data. It is calculated as the product of the standard error and the critical value from the relevant distribution.

The sample size affects the margin of error because a larger sample size reduces the standard error, thereby reducing the margin of error. With a larger sample size, the estimate of the population parameter becomes more precise.

Example: A survey of 50 people gives a margin of error of plus or minus 5%. If the sample size were increased to 200, the margin of error could be reduced to ±2.5%, providing a more precise estimate.

Q10: In a study of the effectiveness of a new weight loss drug, a sample of 50 participants lost an average of 6 pounds with a standard deviation of 2.5 pounds. Conduct a hypothesis test to determine if the drug is significantly effective at a 95% confidence level using a t-test.


```python
import scipy.stats as stats
sample_mean = 6
sample_std_dev = 2.5
sample_size = 50
population_mean = 0
alpha = 0.05
t_statistic = (sample_mean - population_mean) / (sample_std_dev / (sample_size ** 0.5))
df = sample_size - 1
t_critical = stats.t.ppf(1 - alpha / 2, df)
t_statistic, t_critical
```

Q11: In a survey of 500 people, 65% reported being satisfied with their current job. Calculate the 95% confidence interval for the true proportion of people who are satisfied with their job.


```python
sample_size = 500
sample_proportion = 0.65
confidence_level = 0.95
std_error = (sample_proportion * (1 - sample_proportion) / sample_size) ** 0.5
z_critical = stats.norm.ppf(1 - (1 - confidence_level) / 2)
margin_of_error = z_critical * std_error
lower_bound = sample_proportion - margin_of_error
upper_bound = sample_proportion + margin_of_error
(lower_bound, upper_bound)
```

Q12: A researcher is testing the effectiveness of two different teaching methods on student performance. Sample A has a mean score of 85 with a standard deviation of 6, while sample B has a mean score of 82 with a standard deviation of 5. Conduct a hypothesis test to determine if the two teaching methods have a significant difference in student performance using a t-test with a significance level of 0.01.


```python
import scipy.stats as stats
mean_A = 85
std_dev_A = 6
size_A = 30
mean_B = 82
std_dev_B = 5
size_B = 30
alpha = 0.01
pooled_std_error = ((std_dev_A**2 / size_A) + (std_dev_B**2 / size_B)) ** 0.5
t_statistic = (mean_A - mean_B) / pooled_std_error
df = min(size_A - 1, size_B - 1)
t_critical = stats.t.ppf(1 - alpha / 2, df)
t_statistic, t_critical
```

Q13: A population has a mean of 60 and a standard deviation of 8. A sample of 50 observations has a mean of 65. Calculate the 90% confidence interval for the true population mean.


```python
import scipy.stats as stats
population_mean = 60
std_dev = 8
sample_size = 50
sample_mean = 65
confidence_level = 0.90
standard_error = std_dev / (sample_size ** 0.5)
z_critical = stats.norm.ppf(1 - (1 - confidence_level) / 2)
margin_of_error = z_critical * standard_error
confidence_interval = (sample_mean - margin_of_error, sample_mean + margin_of_error)
confidence_interval
```
