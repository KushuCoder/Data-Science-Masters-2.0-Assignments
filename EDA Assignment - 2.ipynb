Q1. Load the flight price dataset and examine its dimensions. How many rows and columns does the dataset have?


```python
import pandas as pd
df = pd.read_csv("flight_price_dataset.csv")
df.shape
```

Q2. What is the distribution of flight prices in the dataset? Create a histogram to visualize the distribution.


```python
import matplotlib.pyplot as plt
plt.hist(df['price'], bins=30, edgecolor='black')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('Distribution of Flight Prices')
plt.show()
```

Q3. What is the range of prices in the dataset? What is the minimum and maximum price?


```python
min_price = df['price'].min()
max_price = df['price'].max()
(min_price, max_price)
```

Q4. How does the price of flights vary by airline? Create a boxplot to compare the prices of different airlines.


```python
import seaborn as sns
df1=df.head()
plt.figure(figsize=(12, 6))
sns.boxplot(x='airline', y='price', data=df1)
plt.xticks(rotation=90)
plt.xlabel('Airline')
plt.ylabel('Price')
plt.title('Flight Price by Airline')
plt.show()
```

Q5. Are there any outliers in the dataset? Identify any potential outliers using a boxplot and describe how they may impact your analysis.


```python
plt.figure(figsize=(12, 6))
sns.boxplot(y=df1['price'])
plt.title('Boxplot of Flight Prices')
plt.show()
```

Q9. Load the Google Playstore dataset and examine its dimensions. How many rows and columns does the dataset have?


```python
df_playstore = pd.read_csv("google_playstore.csv")
df_playstore.shape
```

Q10. How does the rating of apps vary by category? Create a boxplot to compare the ratings of different app categories.


```python
plt.figure(figsize=(12, 6))
sns.boxplot(x='Category', y='Rating', data=df_playstore)
plt.xticks(rotation=90)
plt.xlabel('App Category')
plt.ylabel('Rating')
plt.title('App Ratings by Category')
plt.show()
```

Q11. Are there any missing values in the dataset? Identify any missing values and describe how they may impact your analysis.


```python
df_playstore.isnull().sum()
```

Q12. What is the relationship between the size of an app and its rating? Create a scatter plot to visualize the relationship.


```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x=df_playstore['Size'], y=df_playstore['Rating'])
plt.xlabel('App Size')
plt.ylabel('App Rating')
plt.title('Relationship between App Size and Rating')
plt.show()
```

Q13. How does the type of app affect its price? Create a bar chart to compare average prices by app type.


```python
df_playstore.groupby('Category')['Rating'].mean().plot(kind='bar')
plt.xlabel('App Type')
plt.ylabel('Average Rating')
plt.title('Average Rating of Apps by Type')
plt.show()
```
