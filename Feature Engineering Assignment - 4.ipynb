Q1: What is data encoding? How is it useful in data science?


Data encoding is the process of converting categorical data into numerical format so that machine learning algorithms can process it. It is essential because most algorithms require numerical input to perform calculations. Encoding enables models to interpret and learn from the data effectively.


Q2: What is nominal encoding? Provide an example of how you would use it in a real-world scenario.


Nominal encoding assigns a unique integer to each category in a variable. For example, for a "Color" column with values Red, Green, Blue, the encoded values could be Red=0, Green=1, Blue=2. This method is used when the categorical data has no inherent order, such as product types or city names.


Q3: In what situations is nominal encoding preferred over one-hot encoding? Provide a practical example.


Nominal encoding is preferred over one-hot encoding when the dataset contains high-cardinality categorical variables, meaning a large number of unique categories. For example, in a dataset containing zip codes, one-hot encoding would create too many columns, whereas nominal encoding would convert them into compact numeric codes.


Q4: Suppose you have a dataset containing categorical data with 5 unique values. Which encoding technique would you use to transform this data into a format suitable for machine learning algorithms? Explain why you made this choice.


For a dataset with 5 unique categorical values, one-hot encoding is typically preferred as it avoids implying any ordinal relationship between the categories. However, if the dataset is large and computational efficiency is a concern, nominal encoding may be used instead.


Q5: In a machine learning project, you have a dataset with 1000 rows and 5 columns. Two of the columns are categorical, and the remaining three columns are numerical. If you were to use nominal encoding to transform the categorical data, how many new columns would be created? Show your calculations.


If each of the 2 categorical columns has 4 and 3 unique values respectively, nominal encoding would convert each column into one column of numeric codes, so only 2 new columns would be created. In contrast, one-hot encoding would create (4-1)+(3-1) = 5 new columns.


Q6: You are working with a dataset containing information about different types of animals, including their species, habitat, and diet. Which encoding technique would you use to transform the categorical data into a format suitable for machine learning algorithms? Justify your answer.


For a dataset with species, habitat, and diet, nominal encoding or one-hot encoding can be used. If the number of unique categories is small and there’s no ordinal relationship, one-hot encoding is preferable. For high-cardinality features, nominal encoding is more efficient.
