Q1. What is the difference between Ordinal Encoding and Label Encoding? Provide an example of when you might choose one over the other.


Ordinal Encoding assigns integers to categorical variables based on order or rank. Label Encoding also assigns integers to categories, but it does not consider any particular order among them. 

Use Ordinal Encoding when the categories have a natural order (e.g., low, medium, high). Use Label Encoding when the categories are nominal (e.g., red, blue, green), and no order exists.


Q2. Explain how Target Guided Ordinal Encoding works and provide an example of when you might use it in a machine learning project.


Target Guided Ordinal Encoding assigns numerical values to categorical levels based on the mean of the target variable for each category. 

For example, if you're predicting income based on job title, and average incomes are: Manager - 70k, Clerk - 40k, Executive - 90k, you would encode Executive: 3, Manager: 2, Clerk: 1.


Q3. Define covariance and explain why it is important in statistical analysis. How is covariance calculated?


Covariance is a measure of the relationship between two variables, indicating the direction of their linear relationship. If the covariance is positive, the variables tend to increase together; if negative, one tends to decrease when the other increases.

Covariance between variables X and Y is calculated as:

\ [
\text{Cov}(X, Y) = \frac{1}{n - 1} \sum_{i=1}^n (X_i - \bar{X})(Y_i - \bar{Y})
\ ]


Q4. For a dataset with the following categorical variables: Color (red, green, blue), Size (small, medium, large), and Material (wood, metal, plastic), perform label encoding using Python's scikit-learn library. Show your code and explain the output.


```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
data = pd.DataFrame({
    'Color': ['red', 'green', 'blue'],
    'Size': ['small', 'medium', 'large'],
    'Material': ['wood', 'metal', 'plastic']
})
label_encoders = {}
for column in data.columns:
    le = LabelEncoder()
    data[column] = le.fit_transform(data[column])
    label_encoders[column] = le
print(data)
```

Q5. Calculate the covariance matrix for the following variables in a dataset: Age, Income, and Education level. Interpret the results.


```python
import pandas as pd
import numpy as np
data = pd.DataFrame({
    'Age': [25, 32, 47, 51, 62],
    'Income': [50000, 60000, 75000, 82000, 90000],
    'Education': [12, 16, 16, 18, 20]
})
cov_matrix = np.cov(data.T)
print(cov_matrix)
```

Q7. You are analyzing a dataset with two continuous variables, "Temperature" and "Humidity", and two categorical variables, "Weather Condition" (Sunny/Cloudy/Rainy) and "Wind Direction" (North/South/ East/West). Calculate the covariance between each pair of variables and interpret the results.


```python
import pandas as pd
import numpy as np
data = pd.DataFrame({
    'Temperature': [20, 22, 23, 19, 25],
    'Humidity': [30, 45, 50, 40, 60],
    'Weather Condition': ['Sunny', 'Cloudy', 'Rainy', 'Sunny', 'Rainy'],
    'Wind Direction': ['North', 'South', 'East', 'West', 'North']
})
data_encoded = data.copy()
data_encoded['Weather Condition'] = data_encoded['Weather Condition'].astype('category').cat.codes
data_encoded['Wind Direction'] = data_encoded['Wind Direction'].astype('category').cat.codes
cov_matrix = data_encoded.cov()
print(cov_matrix)
```
