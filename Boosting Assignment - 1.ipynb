Q1. What is boosting in machine learning?


Boosting is an ensemble technique that combines multiple weak learners, typically decision trees, to create a strong predictive model. It trains models sequentially, with each new model attempting to correct the errors of the previous ones.


Q2. What are the advantages and limitations of using boosting techniques?


Advantages:
-High predictive accuracy
-Handles bias well
-Works with various data types

Limitations:
-Sensitive to outliers
-Computationally expensive
-Prone to overfitting if not tuned properly


Q3. Explain how boosting works.


Boosting works by training models sequentially. Each model focuses more on the data points that were misclassified by the previous models. The final prediction is a weighted combination of all models’ outputs.


Q4. What are the different types of boosting algorithms?


-AdaBoost (Adaptive Boosting)
-Gradient Boosting
-XGBoost
-LightGBM
-CatBoost


Q5. What are some common parameters in boosting algorithms?


-'n_estimators': Number of boosting rounds
-'learning_rate': Shrinks the contribution of each tree
-'max_depth': Maximum depth of individual trees
-'subsample': Fraction of samples used for fitting each base learner
-'colsample_bytree': Fraction of features used for each tree


Q6. How do boosting algorithms combine weak learners to create a strong learner?


Boosting algorithms combine weak learners sequentially, with each learner correcting the errors of its predecessor. The final prediction is made by taking a weighted sum (or vote) of the predictions from all learners.


Q7. Explain the concept of AdaBoost algorithm and its working.


AdaBoost works by initially assigning equal weights to all training samples. After each iteration, it increases the weights of the misclassified samples, forcing the next weak learner to focus more on them. The final model is a weighted sum of all weak learners.


Q8. What is the loss function used in AdaBoost algorithm?


AdaBoost uses the exponential loss function, which increases the penalty for misclassified samples exponentially.


Q9. How does the AdaBoost algorithm update the weights of misclassified samples?


In AdaBoost, after each weak learner is trained, the weights of misclassified samples are increased, making them more influential in the next round. Correctly classified samples have their weights decreased.


Q10. What is the effect of increasing the number of estimators in AdaBoost algorithm?


Increasing the number of estimators usually improves performance up to a point, but can also lead to overfitting if too many are used. It also increases training time.
