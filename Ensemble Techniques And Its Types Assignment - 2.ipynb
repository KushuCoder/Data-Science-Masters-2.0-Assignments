Q1. How does bagging reduce overfitting in decision trees?


Bagging reduces overfitting in decision trees by creating multiple subsets of the training data through random sampling with replacement. Each tree is trained on a different subset, and the predictions are averaged (regression) or voted (classification). This reduces variance and makes the final model less sensitive to noise in the training data.


Q2. What are the advantages and disadvantages of using different types of base learners in bagging?


Advantages:
-Flexibility: Bagging can be applied to various models.
-Customization: Allows optimization based on specific data characteristics.

Disadvantages:
-Inconsistent performance: Not all base learners benefit equally from bagging.
-Computational cost: More complex learners increase training time and resource usage.

Q3. How does the choice of base learner affect the bias-variance tradeoff in bagging?


Bagging is most effective with high-variance, low-bias models like decision trees. Using such models reduces variance without increasing bias. If a low-variance, high-bias model is used (e.g., linear regression), bagging may have minimal impact since bias dominates the error.


Q4. Can bagging be used for both classification and regression tasks? How does it differ in each case?


Yes, bagging can be used for both tasks.

-In classification: Each base classifier votes, and the majority class is chosen.
-In regression: The predictions from base regressors are averaged to get the final output.

Q5. What is the role of ensemble size in bagging? How many models should be included in the ensemble?


The ensemble size determines the number of models in the bagging process. Larger ensembles usually improve performance by reducing variance, but after a certain point, returns diminish. A typical choice ranges from 10 to 100 base models, depending on the complexity of the task and computational resources.


Q6. Can you provide an example of a real-world application of bagging in machine learning?


A common real-world application of bagging is in credit scoring. Financial institutions use bagging-based models like Random Forests to predict whether a loan applicant is likely to default. The ensemble approach increases accuracy and robustness by reducing overfitting to the training data.
