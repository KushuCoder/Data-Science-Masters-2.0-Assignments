Q1. Pearson correlation coefficient is a measure of the linear relationship between two variables. Suppose you have collected data on the amount of time students spend studying for an exam and their final exam scores. Calculate the Pearson correlation coefficient between these two variables and interpret the result.


The Pearson correlation coefficient is 1.0, indicating a perfect positive linear relationship between study time and exam scores. This suggests that as the amount of time spent studying increases, the exam scores also increase linearly.


```python
import numpy as np
from scipy.stats import pearsonr
study_time = np.array([2, 4, 6, 8, 10])
exam_scores = np.array([55, 60, 65, 70, 75])
correlation, p_value = pearsonr(study_time, exam_scores)
correlation, p_value
```

Q2. Spearman's rank correlation is a measure of the monotonic relationship between two variables. Suppose you have collected data on the amount of sleep individuals get each night and their overall job satisfaction level on a scale of 1 to 10. Calculate the Spearman's rank correlation between these two variables and interpret the result.


The Spearman's rank correlation coefficient is 1.0, indicating a perfect monotonic relationship. As sleep hours increase, job satisfaction increases consistently.


```python
from scipy.stats import spearmanr
sleep_hours = np.array([6, 7, 8, 5, 9])
job_satisfaction = np.array([4, 6, 7, 3, 8])
correlation, p_value = spearmanr(sleep_hours, job_satisfaction)
correlation, p_value
```

Q3. Suppose you are conducting a study to examine the relationship between the number of hours of exercise per week and body mass index (BMI) in a sample of adults. You collected data on both variables for 50 participants. Calculate the Pearson correlation coefficient and the Spearman's rank correlation between these two variables and compare the results.


Pearson and Spearman correlation coefficients may differ. Pearson checks linearity, while Spearman assesses monotonicity. Both values help understand different aspects of the variable relationship.


```python
exercise_hours = np.random.randint(0, 10, 50)
bmi = np.random.uniform(18, 35, 50)
pearson_corr, _ = pearsonr(exercise_hours, bmi)
spearman_corr, _ = spearmanr(exercise_hours, bmi)
pearson_corr, spearman_corr
```

Q4. A researcher is interested in examining the relationship between the number of hours individuals spend watching television per day and their level of physical activity. The researcher collected data on both variables from a sample of 50 participants. Calculate the Pearson correlation coefficient between these two variables.


The Pearson correlation coefficient here tells us how closely TV watching and physical activity are linearly related. A negative value would indicate that more TV is associated with less activity.


```python
tv_hours = np.random.uniform(1, 5, 50)
physical_activity = np.random.uniform(1, 10, 50)
correlation, p_value = pearsonr(tv_hours, physical_activity)
correlation, p_value
```

Q5. A survey was conducted to examine the relationship between age and preference for a particular brand of soft drink. The survey results are shown below:

| Age (Years) | Soft Drink Preference  |
|-------------|------------------------|
| 25          | Coke                   |
| 42          | Pepsi                  |
| 37          | Mountain dew           |
| 19          | Coke                   |
| 31          | Pepsi                  |
| 28          | Coke                   |


The Spearman's correlation coefficient shows whether a monotonic relationship exists between age and soft drink preference. A near-zero value indicates little to no correlation.


```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from scipy.stats import spearmanr
data = {
    'Age': [25, 42, 37, 19, 31, 28],
    'Soft Drink Preference': ['Coke', 'Pepsi', 'Mountain dew', 'Coke', 'Pepsi', 'Coke']
}
df = pd.DataFrame(data)
le = LabelEncoder()
df['Preference_Encoded'] = le.fit_transform(df['Soft Drink Preference'])
correlation, p_value = spearmanr(df['Age'], df['Preference_Encoded'])
correlation, p_value
```
