Linear Regression is used for predicting continuous values. It models the relationship between input variables (X) and a continuous output (y) using a linear equation.    

Logistic Regression is used for classification tasks (typically binary classification). It predicts the probability that an input belongs to a certain class using the sigmoid function to map predictions to values between 0 and 1.

Q2. What is the cost function used in logistic regression, and how is it optimized?


In logistic regression, the cost function used is called log loss or binary cross-entropy. It measures the difference between the predicted probabilities and the actual class labels.     
J(theta) = -1/m * sum ( y * log(h_theta(x)) + (1 - y) * log(1 - h_theta(x)) )    This cost function is optimized using gradient descent, which adjusts the model's parameters step by step to minimize the cost. Other optimization techniques like stochastic gradient descent (SGD) or LBFGS can also be used.


Q3. Explain the concept of regularization in logistic regression and how it helps prevent overfitting.


Regularization adds a penalty to the cost function to discourage overly complex models, helping to reduce overfitting.    

Two common types:    

L1 (Lasso): Encourages sparsity by shrinking some coefficients to zero (feature selection).    

L2 (Ridge): Penalizes large coefficients, making the model more robust.

Q4. What is the ROC curve, and how is it used to evaluate the performance of the logistic regression
model?


The ROC (Receiver Operating Characteristic) Curve plots:    

True Positive Rate (Recall) vs. False Positive Rate at various threshold settings.    

It helps visualize the model’s performance across different classification thresholds.    

The Area Under the Curve (AUC) quantifies overall performance:    

AUC = 1.0: Perfect classifier    

AUC = 0.5: No better than random guessing

Q6. How can you handle imbalanced datasets in logistic regression? What are some strategies for dealing
with class imbalance?


Strategies:    

Resampling:     

Oversampling the minority class (e.g., SMOTE)    

Undersampling the majority class    

Class weights: Assign higher penalty to misclassifying minority class using class_weight='balanced' in sklearn    

Use different evaluation metrics: Precision, Recall, F1-score, AUC-ROC (instead of accuracy)    



Q7. Can you discuss some common issues and challenges that may arise when implementing logistic
regression, and how they can be addressed? For example, what can be done if there is multicollinearity
among the independent variables?

Multicollinearity:
Happens when features are highly correlated.    
Fixes:    
Remove one of the correlated variables    
Use Principal Component Analysis (PCA)    
Apply L2 regularization    
Outliers:
Can distort predictions.    
Solution: Use robust techniques or transform the features.    
Non-linear relationships:
Logistic regression assumes a linear decision boundary.    
Fix: Add polynomial or interaction terms, or use non-linear models.    
Missing data:
Handle via imputation or removal depending on the situation.
