Q1. What is the purpose of grid search cv in machine learning, and how does it work?


Grid Search CV (Cross-Validation) is used in machine learning to find the best combination of hyperparameters for a model. It works by performing an exhaustive search over a specified parameter grid and evaluates each combination using cross-validation to select the best model configuration.

This helps in improving model performance by tuning the model to its optimal set of hyperparameters.


Q2. Describe the difference between grid search cv and randomize search cv, and when might you choose one over the other?


Grid Search CV searches all possible combinations of hyperparameters in the provided grid, while Randomized Search CV selects a fixed number of random combinations to evaluate from the grid.

Choose Grid Search CV when the grid is small and exhaustive search is feasible. Choose Randomized Search CV when the parameter space is large and computational resources are limited.


Q3. What is data leakage, and why is it a problem in machine learning? Provide an example.


Data leakage occurs when information from outside the training dataset is used to create the model. This can lead to overly optimistic performance estimates and poor generalization to new data.

Example: Using a feature that is only available after prediction (like future sales data) to train the model.

Q5. What is a confusion matrix, and what does it tell you about the performance of a classification model?


A confusion matrix is a table used to evaluate the performance of a classification model. It shows the counts of actual versus predicted classifications, broken down into True Positives, False Positives, True Negatives, and False Negatives.

It helps identify the types of errors the model makes.


Q6. Explain the difference between precision and recall in the context of a confusion matrix.


Precision is the proportion of true positive predictions among all positive predictions:

Precision = TP / (TP + FP)

Recall is the proportion of true positives identified out of all actual positives:

Recall = TP / (TP + FN)


Q7. How can you interpret a confusion matrix to determine which types of errors your model is making?


By analyzing the values in a confusion matrix:

-High FP: Model often wrongly labels negatives as positives
-High FN: Model misses many actual positives

This helps decide whether precision or recall is more important based on the context.


Q8. What are some common metrics that can be derived from a confusion matrix, and how are they calculated?


Common metrics:

-Accuracy = (TP + TN) / (TP + TN + FP + FN)
-Precision = TP / (TP + FP)
-Recall = TP / (TP + FN)
-F1 Score = 2 * (Precision * Recall) / (Precision + Recall)

These metrics provide a more complete evaluation than accuracy alone.


Q9. What is the relationship between the accuracy of a model and the values in its confusion matrix?


Accuracy is calculated using the confusion matrix values: (TP + TN) / Total predictions. A high accuracy might be misleading in imbalanced datasets, so it's important to consider other metrics as well.


Q10. How can you use a confusion matrix to identify potential biases or limitations in your machine learning model?


A confusion matrix can reveal if a model performs poorly on certain classes, which could indicate class imbalance or biased data. By analyzing false positives and false negatives, you can identify weaknesses in model predictions and make improvements.
