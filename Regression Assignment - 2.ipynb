Q1. Explain the concept of R-squared in linear regression models. How is it calculated, and what does it represent?


R-squared, also known as the coefficient of determination, measures the proportion of variance in the dependent variable that is explained by the independent variables in a regression model. It is calculated as:

R² = 1 - (SS_res / SS_tot)

where:
-SS_res is the sum of squared residuals (errors)
-SS_tot is the total sum of squares

R-squared ranges from 0 to 1, where a value closer to 1 indicates a better fit of the model to the data.


Q2. Define adjusted R-squared and explain how it differs from the regular R-squared.


Adjusted R-squared modifies the R-squared value to account for the number of predictors in the model. Unlike R-squared, which always increases when adding more predictors, adjusted R-squared only increases if the new predictor improves the model significantly. It is calculated as:

Adjusted R² = 1 - [(1 - R²) * (n - 1) / (n - p - 1)]

where:
-n is the number of observations
-p is the number of predictors

Adjusted R-squared provides a more accurate measure of model performance, especially when dealing with multiple predictors.


Q3. When is it more appropriate to use adjusted R-squared?


Adjusted R-squared is more appropriate when comparing models with different numbers of predictors. It prevents overestimation of the model's performance that can occur when additional, non-informative predictors are added.


Q4. What are RMSE, MSE, and MAE in the context of regression analysis? How are these metrics calculated, and what do they represent?


RMSE (Root Mean Squared Error), MSE (Mean Squared Error), and MAE (Mean Absolute Error) are common regression evaluation metrics:

-MSE: The average squared difference between actual and predicted values.
-RMSE: The square root of MSE, which provides error in the same units as the dependent variable.
-MAE: The average absolute difference between actual and predicted values.

These metrics measure the accuracy of a regression model, with lower values indicating better performance.


Q5. Discuss the advantages and disadvantages of using RMSE, MSE, and MAE as evaluation metrics in regression analysis.


Advantages and disadvantages of RMSE, MSE, and MAE:

-RMSE:
  -Advantage: Penalizes large errors more, making it useful for models where large deviations are critical.
  -Disadvantage: Sensitive to outliers.

-MSE:
  -Advantage: Smooth and differentiable, making it useful for optimization.
  -Disadvantage: Difficult to interpret because it squares errors.

-MAE:
  -Advantage: Provides a simple interpretation as the average error.
  -Disadvantage: Less sensitive to large errors compared to RMSE.


Q6. Explain the concept of Lasso regularization. How does it differ from Ridge regularization, and when is it more appropriate to use?


Lasso (Least Absolute Shrinkage and Selection Operator) regularization adds an L1 penalty to the regression model:

Lasso Loss = RSS + λ * Σ|β_j|

where:
-RSS is the residual sum of squares
-λ is the regularization parameter
-β_j are the regression coefficients

Lasso differs from Ridge regularization (which uses L2 penalty) because it can shrink some coefficients to zero, effectively performing feature selection. It is useful when dealing with high-dimensional data where feature selection is needed.


Q7. How do regularized linear models help to prevent overfitting in machine learning? Provide an example to illustrate.


Regularized linear models prevent overfitting by adding a penalty to large coefficients, reducing model complexity. For example:

-Ridge regression shrinks coefficients but keeps all variables.
-Lasso regression shrinks some coefficients to zero, removing less relevant features.

This helps create more generalizable models that perform well on unseen data.

Q8. Discuss the limitations of regularized linear models and explain why they may not always be the best choice for regression analysis.


Limitations of regularized linear models:

-They assume that all features are equally important, which may not always be true.
-Regularization parameters (λ) require tuning, which adds computational complexity.
-Lasso regression can struggle when dealing with highly correlated features, arbitrarily selecting one feature over another.


Q9. You are comparing the performance of two regression models using different evaluation metrics.
Model A has an RMSE of 10, while Model B has an MAE of 8. Which model would you choose as the better performer, and why? Are there any limitations to your choice of metric?


Choosing the better model between Model A (RMSE = 10) and Model B (MAE = 8) depends on the problem:

-RMSE penalizes large errors more than MAE, making it suitable for cases where large errors are critical.
-MAE provides a more interpretable measure of average error.

If minimizing large deviations is important, Model A might be preferred. However, if overall consistency is key, Model B could be a better choice.


Q10. You are comparing the performance of two regularized linear models using different types of regularization. Model A uses Ridge regularization with a regularization parameter of 0.1, while Model B uses Lasso regularization with a regularization parameter of 0.5. Which model would you choose as the better performer, and why? Are there any trade-offs or limitations to your choice of regularization method?


Choosing between Ridge (Model A) and Lasso (Model B):

-Ridge regression (λ = 0.1) is better when all features contribute to prediction and multicollinearity is present.
-Lasso regression (λ = 0.5) is better for feature selection when some predictors are irrelevant.

Trade-offs:
-Ridge retains all features but shrinks them.
-Lasso removes irrelevant features but may eliminate useful correlated features.

The choice depends on whether feature selection or retaining all predictors is preferred.
