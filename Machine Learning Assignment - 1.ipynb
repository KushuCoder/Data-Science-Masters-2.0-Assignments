Q1: Explain the following with an example:
   1 ) Artificial Intelligence  
   2 ) Machine Learning  
   3 ) Deep Learning  


 Artificial Intelligence (AI) is the broader concept of machines being able to perform tasks in a way that we would consider "smart." It includes any technique that enables computers to mimic human intelligence.

Machine Learning (ML) is a subset of AI that allows computers to learn from data and make predictions or decisions without being explicitly programmed.

Deep Learning (DL) is a subset of ML that uses neural networks with multiple layers to process and learn from large amounts of data.

Example:  
-AI: A chatbot that answers customer inquiries.  
-ML: A recommendation system that suggests products based on past purchases.  
-DL: A self-driving car detecting objects and making decisions in real-time.


Q2: What is supervised learning? List some examples of supervised learning.


 Supervised learning is a type of machine learning where the model is trained on labeled data. The algorithm learns the relationship between input and output variables and makes predictions on new data.

Examples of supervised learning:  
-Spam detection in emails  
-Credit scoring in banking  
-Image classification (cat vs. dog)  
-Speech recognition  


Q3: What is unsupervised learning? List some examples of unsupervised learning.


 Unsupervised learning is a type of machine learning where the model is trained on unlabeled data. The algorithm identifies patterns, relationships, or structures within the data without explicit labels.

Examples of unsupervised learning:  
-Customer segmentation in marketing  
-Anomaly detection in fraud detection  
-Clustering similar documents  
-Dimensionality reduction techniques like PCA  


Q4: What is the difference between AI, ML, DL, and DS?


AI (Artificial Intelligence) is the overall concept of machines mimicking human intelligence.  
ML (Machine Learning) is a subset of AI that enables machines to learn from data.  
DL (Deep Learning) is a subset of ML that uses neural networks for complex learning tasks.  
DS (Data Science) is a field that uses statistics, ML, and data analysis techniques to extract insights from data.  

Example:  
-AI: A virtual assistant like Siri  
-ML: A spam email filter  
-DL: Facial recognition on smartphones  
-DS: Analyzing customer behavior trends  


Q5: What are the main differences between supervised, unsupervised, and semi-supervised learning?


Supervised learning uses labeled data and is used for prediction tasks.  
Unsupervised learning works with unlabeled data and is used for discovering patterns.  
Semi-supervised learning is a mix of both, where a small amount of labeled data is combined with a large amount of unlabeled data.  

Example:  
-Supervised: Predicting house prices  
-Unsupervised: Grouping customers based on buying behavior  
-Semi-supervised: Fraud detection where only a few cases are labeled  


Q6: What is train, test and validation split? Explain the importance of each term.


Train, test, and validation split is essential in machine learning to evaluate model performance properly.  

-Training set: Used to train the model and learn patterns.  
-Validation set: Used to fine-tune hyperparameters and prevent overfitting.  
-Test set: Used to evaluate the final model performance on unseen data.  

A typical split is 70% training, 15% validation, and 15% testing.  


Q7: How can unsupervised learning be used in anomaly detection?


Unsupervised learning can be used in anomaly detection by identifying patterns and deviations from those patterns. Clustering and density-based methods can detect outliers.

Example techniques:  
-K-Means clustering: Groups normal data points, and outliers are far from clusters.  
-DBSCAN: Identifies dense areas, marking sparse regions as anomalies.  
-Autoencoders: Learn to reconstruct normal data, and high reconstruction error indicates anomalies.
