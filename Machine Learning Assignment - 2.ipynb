Q1: Define overfitting and underfitting in machine learning. What are the consequences of each, and how can they be mitigated?


Overfitting occurs when a machine learning model learns the noise in the training data instead of the underlying pattern. This leads to high accuracy on the training set but poor generalization to new data. Underfitting happens when the model is too simple to capture the patterns in the data, resulting in poor performance on both training and test data.

Consequences:  
-Overfitting: Poor generalization, high variance, complex models.
-Underfitting: High bias, poor performance on training and test sets.

Mitigation:  
-Overfitting: Regularization, more training data, dropout, pruning.
-Underfitting: More complex models, feature engineering, reducing regularization.


Q2: How can we reduce overfitting? Explain in brief.


Overfitting can be reduced using several techniques:

-Cross-validation: Helps estimate model performance on unseen data.
-Regularization: Methods like L1 (Lasso) and L2 (Ridge) prevent overly complex models.
-Pruning: In decision trees, pruning reduces complexity by removing less significant branches.
-Increasing data: More diverse training data helps the model generalize better.
-Dropout: In neural networks, dropout randomly removes some neurons during training to prevent dependency on specific features.


Q3: Explain underfitting. List scenarios where underfitting can occur in ML.


Underfitting occurs when a model is too simple to learn the underlying structure in the data. It leads to high bias and poor predictive performance.

Scenarios where underfitting can occur:  
-Using a linear model on non-linear data.
-Insufficient training due to very few iterations in optimization.
-Excessive regularization, preventing the model from capturing necessary patterns.
-Poor feature selection, where important features are ignored.


Q4: Explain the bias-variance tradeoff in machine learning. What is the relationship between bias and variance, and how do they affect model performance?


The bias-variance tradeoff refers to the balance between a model's ability to generalize to new data and its accuracy on training data.

-High bias: The model makes strong assumptions, leading to underfitting.
-High variance: The model is too sensitive to training data, leading to overfitting.

A good model maintains a balance where both bias and variance are minimized.


Q5: Discuss some common methods for detecting overfitting and underfitting in machine learning models. How can you determine whether your model is overfitting or underfitting?


Methods for detecting overfitting and underfitting:

-Train-test performance comparison: If training accuracy is high and test accuracy is low, the model is overfitting.
-Learning curves: If both training and validation errors are high, the model is underfitting. If training error is low and validation error is high, the model is overfitting.
-Cross-validation: Helps assess model performance on unseen data.
-Regularization impact: Increasing regularization reduces overfitting but might lead to underfitting if too strong.


Q6: Compare and contrast bias and variance in machine learning. What are some examples of high bias and high variance models, and how do they differ in terms of their performance?


Bias refers to systematic errors due to incorrect assumptions in the model. Variance refers to sensitivity to training data fluctuations.

-High bias models: Linear regression on non-linear data, decision trees with shallow depth.
-High variance models: Deep neural networks without regularization, decision trees with large depth.

High bias models fail to capture patterns, while high variance models overfit training data.


Q7: What is regularization in machine learning, and how can it be used to prevent overfitting? Describe some common regularization techniques and how they work.


 Regularization is a technique used to reduce overfitting by adding a penalty term to the loss function, discouraging overly complex models.

Common regularization techniques:
-L1 Regularization (Lasso): Shrinks some feature coefficients to zero, promoting sparsity.
-L2 Regularization (Ridge): Penalizes large coefficients, reducing model complexity.
-Dropout: In neural networks, randomly drops some neurons to prevent reliance on specific features.
